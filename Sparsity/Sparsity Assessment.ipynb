```python
import os
import json
import numpy as np

folder_path = "attention_matrices"
threshold = 1e-3
sparsity_info = []

for filename in os.listdir(folder_path):
    if filename.endswith(".json"):
        file_path = os.path.join(folder_path, filename)
        with open(file_path, 'r') as f:
            data = json.load(f)
        
        if isinstance(data, dict):
            if "mean_attention_final_step" in data:
                data = np.array(data["mean_attention_final_step"])
            else:
                continue  
        else:
            data = np.array(data)
        
        num_zero = np.sum(data < threshold)
        sparsity = num_zero / data.size
        sparsity_info.append((filename, sparsity))

sparsity_info.sort(key=lambda x: x[1], reverse=True)

for filename, sparsity in sparsity_info:
    print(f"{filename}: sparsity = {sparsity:.2f}")
```

```python
import matplotlib.pyplot as plt

sparsity_info.sort(key=lambda x: x[1], reverse=True)
filenames, sparsities = zip(*sparsity_info)

plt.rcParams.update({
    'font.size': 24,         
    'axes.titlesize': 24,    
    'axes.labelsize': 24,    
    'xtick.labelsize': 13,   
    'ytick.labelsize': 13    
})

# bar graph
plt.figure(figsize=(12, 6))
plt.bar(range(len(sparsities)), sparsities, color='skyblue')
plt.xticks()
plt.title('Sparsity of Attention Matrices')
plt.tight_layout()
plt.show()

# histogram
plt.figure(figsize=(8, 5))
plt.hist(sparsities, bins=20, color='salmon', edgecolor='black')

plt.xlim(0.4, 1.0)

plt.xlabel('Sparsity')
plt.title('Distribution of Sparsity Across Attention Matrices')
plt.tight_layout()
plt.show()
```

```python
import json
import numpy as np
import matplotlib.pyplot as plt

file_path = "attention_matrices/attn_0117.json"  
threshold = 1e-3

with open(file_path, "r") as f:
    data = json.load(f)

matrix = np.array(data["mean_attention_final_step"])
matrix = np.squeeze(matrix)  

# calculate sparsity per step
sparsity_per_step = np.mean(matrix < threshold, axis=1)

plt.figure(figsize=(8, 4))
plt.plot(sparsity_per_step, marker='o')
plt.title("Attention Sparsity Over Time")
plt.xlabel("Generation Step")
plt.ylabel("Sparsity (Fraction Near Zero)")
plt.ylim(0, 1)
plt.grid(True)
plt.show()
```
