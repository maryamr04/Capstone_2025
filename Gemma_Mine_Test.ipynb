```python
from google.colab import drive, files
import pandas as pd
```

```python
drive.mount('/content/drive')
df = pd.read_csv("/content/drive/MyDrive/Capstone/Data/mines.csv")
```

```python
from transformers import pipeline
from huggingface_hub import login
from tqdm import tqdm

login("YOUR-KEY")

pipe = pipeline(
    task="text-generation",
    model="google/gemma-3-4b-it",  #google/gemma-3-4b-it
    padding=True,
    truncation=True,
    device_map="auto"
)

recommendations = []
for _, row in tqdm(df.iterrows(), total=1, desc="Recommending sensors"):
    context = str(row["Context"])
    target = str(row["Target"])

    prompt = (
        f"You are a sensor expert. Given the following situation:\n"
        f"Context: {context}\n"
        f"Target: {target}\n\n"
        f"Recommend the most appropriate type of sensor to use and briefly explain why."
    )

    output = pipe(prompt, max_new_tokens=100, temperature=0.7)
    rec = output[0]["generated_text"]
    recommendations.append(rec)

df["sensor_recommendation"] = recommendations
df.to_csv("sensor_recommendations.csv", index=False)
```
